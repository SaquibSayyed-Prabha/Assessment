### PART 3 : Data Cleaning

```python
# 1. Find missing values. 
print(df.isnull())
```

```python
# 2. Count missing values. 
print(df.isnull().sum())
```

```python
# 3. Fill missing values in Age using the average age. 
avg_age = df["Age"].mean()
df["Age"] = df["Age"].fillna(avg_age)

print(df)
```

```python
# 4. Fill missing values in Marks using the average marks. 
avg_marks = df["Marks"].mean()
df["Marks"] = df["Marks"].fillna(avg_marks)

print(df)
```

```python
# 5. Fill missing values in Attendance using the average attendance. 
avg_attendance = df["Attendance"].mean()
df["Attendance"] = df["Attendance"].fillna(avg_attendance)

print(df)
```

```python
# 6. Verify that no missing values exist. 
print(df.isnull())
```

### PART 11 : Aggregation Functions 
Using groupby() and agg(), display the following for each Course:
1. Sum of Marks. 
2. Average Marks. 
3. Minimum Marks. 
4. Maximum Marks. 
5. Total Students. 

```python
print(df.groupby("Course").agg(
    Total_Marks=("Marks", "sum"),
    Average_Marks=("Marks", "mean"),
    Minimum_Marks=("Marks", "min"),
    Maximum_Marks=("Marks", "max"),
    Total_Students=("Student_ID", "count")
))
```

### PART 13 : Handling Categorical Data 
Perform Manual Encoding:

```python
# Gender 
# Male = 1 
# Female = 0 

df["Gender"] = df["Gender"].map({
    "Male": 1,
    "Female": 0
})

print(df)
```

```python
# Purchased 
# Yes = 1 
# No = 0 

df["Purchased"] = df["Purchased"].map({
    "Yes": 1,
    "No": 0
})

print(df)
```
